```python
import pandas as pd
import numpy as np

# Memuat dataset hasil Rule-Based Detection
df = pd.read_csv(
    r"C:\AI-Anomaly-Detection\outputs\hasil_rule_based_detection.csv"
)

# Mengubah kolom waktu menjadi datetime
df["waktu_tercatat"] = pd.to_datetime(
    df["waktu_tercatat"]
)

# Mengurutkan data berdasarkan sesi dan waktu
df = df.sort_values(
    ["session_id", "waktu_tercatat"]
).reset_index(drop=True)

print("Jumlah baris:", len(df))
print("Jumlah kolom:", len(df.columns))

print("\nDaftar kolom:")
print(df.columns.tolist())
```

```python
print("===== INFORMASI DATASET =====")
df.info()

print("\n===== 5 DATA PERTAMA =====")
display(df.head())

print("\n===== DISTRIBUSI RULE-BASED =====")
print(df["rule_based_anomaly"].value_counts())
```

```python
# Fitur sensor utama
sensor_features = [
    "suhu",
    "kelembaban",
    "pm10",
    "pm25",
    "barometer",
    "tvoc"
]

# Fitur temporal
temporal_features = [
    "tvoc_duration",
    "tvoc_change_flag",
    "tvoc_delta",
    "barometer_duration",
    "barometer_change_flag",
    "barometer_delta",
    "pm10_duration",
    "pm10_change_flag",
    "pm10_delta",
    "pm25_duration",
    "pm25_change_flag",
    "pm25_delta",
    "gap_panjang_flag",
    "awal_sesi_flag"
]

# Menggabungkan fitur
feature_columns = sensor_features + temporal_features

# Membuat dataset fitur
X = df[feature_columns].copy()

print("Jumlah fitur:", len(feature_columns))
print("Jumlah data:", len(X))

print("\nDaftar fitur:")
print(feature_columns)

print("\nPemeriksaan nilai kosong:")
print(X.isnull().sum())
```

```python
# Mengambil daftar sesi sesuai urutan kemunculannya
daftar_sesi = sorted(df["session_id"].unique())

# Membagi sesi 80:20
split_index = int(len(daftar_sesi) * 0.8)

sesi_train = daftar_sesi[:split_index]
sesi_test = daftar_sesi[split_index:]

# Membuat pembagian data
mask_train = df["session_id"].isin(sesi_train)
mask_test = df["session_id"].isin(sesi_test)

X_train = X.loc[mask_train].copy()
X_test = X.loc[mask_test].copy()

print("Jumlah seluruh sesi:", len(daftar_sesi))
print("Jumlah sesi training:", len(sesi_train))
print("Jumlah sesi testing:", len(sesi_test))

print("\nJumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))

print("\nSesi training:", sesi_train)
print("Sesi testing:", sesi_test)

# Memastikan tidak ada sesi yang tumpang tindih
print("\nSesi tumpang tindih:",
      set(sesi_train).intersection(set(sesi_test)))
```

```python
from sklearn.ensemble import IsolationForest

# Membuat model Isolation Forest
model_if = IsolationForest(
    n_estimators=200,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

# Melatih model hanya menggunakan data training
model_if.fit(X_train)

print("Pelatihan Isolation Forest berhasil!")
print("Jumlah data training:", len(X_train))
print("Jumlah fitur:", X_train.shape[1])
```

```python
# Melakukan prediksi pada data testing
prediksi_if = model_if.predict(X_test)

# Menghitung anomaly score
skor_if = model_if.decision_function(X_test)

# Menyimpan hasil prediksi
hasil_testing = df.loc[X_test.index].copy()

hasil_testing["prediksi_if"] = prediksi_if
hasil_testing["skor_if"] = skor_if

# Mengubah hasil prediksi menjadi label
hasil_testing["ai_anomaly"] = (
    hasil_testing["prediksi_if"] == -1
)

print("===== HASIL PREDIKSI ISOLATION FOREST =====")

print("\nDistribusi prediksi:")
print(
    hasil_testing["ai_anomaly"]
    .value_counts()
    .rename(index={True: "Anomali", False: "Normal"})
)

print("\nPersentase prediksi:")
print(
    hasil_testing["ai_anomaly"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nContoh hasil prediksi:")
display(
    hasil_testing[
        [
            "session_id",
            "waktu_tercatat",
            "pm10",
            "pm25",
            "tvoc",
            "skor_if",
            "ai_anomaly"
        ]
    ].head(10)
)
```

```python
# Menggabungkan hasil deteksi Rule-Based dan AI
hasil_testing["rule_based_anomaly"] = (
    hasil_testing["rule_based_anomaly"].astype(bool)
)

hasil_testing["ai_anomaly"] = (
    hasil_testing["ai_anomaly"].astype(bool)
)

# Menghitung kombinasi hasil deteksi
kondisi = [
    (
        hasil_testing["rule_based_anomaly"] &
        hasil_testing["ai_anomaly"]
    ),
    (
        hasil_testing["rule_based_anomaly"] &
        ~hasil_testing["ai_anomaly"]
    ),
    (
        ~hasil_testing["rule_based_anomaly"] &
        hasil_testing["ai_anomaly"]
    ),
    (
        ~hasil_testing["rule_based_anomaly"] &
        ~hasil_testing["ai_anomaly"]
    )
]

kategori = [
    "Rule-Based + AI",
    "Rule-Based saja",
    "AI saja",
    "Normal menurut keduanya"
]

hasil_testing["perbandingan_deteksi"] = np.select(
    kondisi,
    kategori,
    default="Tidak diketahui"
)

# Menampilkan jumlah setiap kombinasi
print("===== PERBANDINGAN RULE-BASED DAN AI =====")

print(
    hasil_testing["perbandingan_deteksi"]
    .value_counts()
)

print("\n===== PERSENTASE =====")

print(
    hasil_testing["perbandingan_deteksi"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# Menentukan kategori Hybrid AI
kondisi_hybrid = [
    (
        hasil_testing["rule_based_anomaly"] &
        hasil_testing["ai_anomaly"]
    ),
    (
        hasil_testing["rule_based_anomaly"] &
        ~hasil_testing["ai_anomaly"]
    ),
    (
        ~hasil_testing["rule_based_anomaly"] &
        hasil_testing["ai_anomaly"]
    )
]

kategori_hybrid = [
    "Anomali - Rule-Based + AI",
    "Anomali - Rule-Based saja",
    "Anomali - AI saja"
]

hasil_testing["kategori_hybrid"] = np.select(
    kondisi_hybrid,
    kategori_hybrid,
    default="Normal"
)

# Penanda kandidat anomali Hybrid
hasil_testing["hybrid_anomaly"] = (
    hasil_testing["rule_based_anomaly"] |
    hasil_testing["ai_anomaly"]
)

# Menampilkan distribusi hasil Hybrid
print("===== HASIL HYBRID AI =====")

print(
    hasil_testing["kategori_hybrid"]
    .value_counts()
)

print("\n===== TOTAL KANDIDAT ANOMALI HYBRID =====")

print(
    hasil_testing["hybrid_anomaly"]
    .value_counts()
)
```

```python
# Menghitung jenis kandidat anomali dari Rule-Based
kolom_kategori = [
    "kandidat_spike",
    "spike_reversal",
    "stuck_value",
    "unusual_value"
]

print("===== RINCIAN RULE-BASED PADA DATA TESTING =====")

for kolom in kolom_kategori:
    jumlah = hasil_testing[kolom].sum()

    print(f"{kolom}: {jumlah}")

print("\n===== RINCIAN KATEGORI HYBRID =====")

display(
    hasil_testing[
        hasil_testing["hybrid_anomaly"]
    ]["kategori_hybrid"]
    .value_counts()
    .to_frame("jumlah_data")
)
```

```python
print("===== KOMBINASI RULE-BASED PADA DATA TESTING =====")

display(
    hasil_testing["kombinasi_deteksi"]
    .value_counts()
    .to_frame("jumlah_data")
)

print("\n===== TUMPANG TINDIH JENIS ANOMALI =====")

display(
    pd.crosstab(
        hasil_testing["stuck_value"],
        hasil_testing["unusual_value"],
        margins=True
    )
)
```
